## Scenario

The pathologist wants to understand the characteristics of the tumours before comparing different patient groups.

## Task

Using the cleaned patient dataset, create three plots exploring pathological features:

1. Visualize the distribution of tumour grades.
2. Visualize the distribution of tumour stages.
3. Visualize the relationship between tumour grade and tumour stage.

```python
# import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# assess data for relevant pathological variables
df = pd.read_csv("challenge5.csv")  # replace if the filename differs

print(df.head())
print(df.columns)
```

```python
# univariate analysis: tumour grade distribution
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="Sample Grade")

plt.title("Tumour Grade Distribution")
plt.xlabel("Tumour Grade")
plt.ylabel("Number of Patients")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# univariate analysis: tumour stage distribution

plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="Sample Stage")

plt.title("Tumour Stage Distribution")
plt.xlabel("Tumour Stage")
plt.ylabel("Number of Patients")
plt.tight_layout()
plt.show()
```

```python
# relationship between tumour grade and tumour stage

plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="Sample Stage", hue="Sample Grade")

plt.title("Tumour Grade by Tumour Stage")
plt.xlabel("Tumour Stage")
plt.ylabel("Number of Patients")
plt.legend(title="Tumour Grade")
plt.tight_layout()
plt.show()
```
