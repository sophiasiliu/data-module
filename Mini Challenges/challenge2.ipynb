```python
# import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# assess data for relevant variables & relationships
df = pd.read_csv("chellenge2_data.csv") #TODO: replace with correct file
print(df.head())
print(df.columns)
```

```python
#univariate analysis #TODO: replace variable?
#Patient Age at Diagnosis
bins = [0, 40, 50, 60, 70, 80, 100]
labels = ["<40", "40-49", "50-59", "60-69", "70-79", "80+"]

# creates new column "Age Group" with age groups based on "Patient Age Diagnosis"
df["Age Group"] = pd.cut(df["Patient Age Diagnosis"], bins=bins, labels=labels, right=False)

# get value counts of each age group
age_counts = df["Age Group"].value_counts().sort_index()

# plot
plt.figure(figsize=(10, 6))
age_counts.plot(kind="bar")

plt.title("Patient Age at Diagnosis Distribution")
plt.xlabel("Age")
plt.ylabel("Count")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
#multivariate analysis
# Sample Stage vs. Age  #TODO: replace variables?
plt.figure(figsize=(10, 6))

# make violin plot of tumour subtype vs age
sns.violinplot(data=df, x="Sample Stage", y="Patient Age Diagnosis")
plt.title("Sample Stage vs Age Distribution")
plt.xlabel("Sample Stage")
plt.ylabel("Age")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
