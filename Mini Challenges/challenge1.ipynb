```python
# import libraries
import pandas as pd
import numpy as np
```

```python
# check data for preprocessing steps that need to be applied
df = pd.read_csv("challenge1_data.csv")
print(df.head())
print(df.columns)

# check data types of each column
print(df.dtypes)

# check which features are missing values
print(df.isnull().sum()) # missing in only death_date column so can proceed without dropping any rows

# check duplicates
print(df.duplicated().sum()) # no duplicates! can proceed
```

```python
# convert certain columns to integer
df["age"] = pd.to_numeric(df["age"], errors='coerce').astype("Int64")

# convert date columns to datetime
df['diagonsis_date'] = pd.to_datetime(df['diagonsis_date'], errors='coerce')
df['surgery_date'] = pd.to_datetime(df['surgery_date'], errors='coerce')
df['death_date'] = pd.to_datetime(df['death_date'], errors='coerce')

# check data types after
print(df.dtypes)
```

```python
# check values in df columns for mixed data types and values
for column in df.columns:
    print(f"Unique values in {column}: {df[column].unique()}")

# notice certain columns have mixed data types and values, so we need to clean them up
# hospital, diagnosis, sex, stage, grade, subtype, tumour_size, er_status, pr_status, her2_status
```

```python
# edit column values to make them consistent

# hospital: ['Kingston General' "St Mary's" "Queen's Hospital" 'St. Marys']
hospital_mapping = {"St. Marys": "St Mary's", "Queen's Hospital": "Kingston General"}
df["hospital"] = df["hospital"].replace(hospital_mapping)

# diagnosis: ['Breast Cancer' 'breast cancer' 'BREAST CANCER' 'Breast cancer', 'breast Cancer' 'BREAST cancer']
diagnosis_mapping = {"breast cancer": "Breast Cancer", "BREAST CANCER": "Breast Cancer", "Breast cancer": "Breast Cancer", "breast Cancer": "Breast Cancer", "BREAST cancer": "Breast Cancer"}
df["diagnosis"] = df["diagnosis"].replace(diagnosis_mapping)

# sex: ['Female' 'female' 'F']
sex_mapping = {"female": "Female", "F": "Female"}
df["sex"] = df["sex"].replace(sex_mapping)

# stage: ['II' 'III' '1' 'Stage IV' 'I' '3' 'IV' 'Stage III' '2']
stage_mapping = {"II": "Stage 2", "III": "Stage 3", "1": "Stage 1", "Stage IV": "Stage 4", "I": "Stage 1", "3": "Stage 3", "IV": "Stage 4", "Stage III": "Stage 3", "2": "Stage 2"}
df["stage"] = df["stage"].replace(stage_mapping)

# grade: ['2' 'Grade 3' 'G1' 'Grade 2' '1' '3' 'G2' 'Grade 1' 'G3']
grade_mapping = {"G1": "Grade 1", "G2": "Grade 2", "G3": "Grade 3",  "1": "Grade 1", "2": "Grade 2", "3": "Grade 3"}
df["grade"] = df["grade"].replace(grade_mapping)

# subtype: ['Invasive Ductal' 'invasive ductal carcinoma' 'IDC' 'Invasive Lobular', 'Lobular' 'Ductal' 'Invasive ductal' 'ILC' 'Invasive ductal carcinoma']
subtype_mapping = {"invasive ductal carcinoma": "Invasive Ductal", "IDC": "Invasive Ductal", "Lobular": "Invasive Lobular", "Ductal": "Invasive Ductal", "Invasive ductal": "Invasive Ductal", "ILC": "Invasive Lobular", "Invasive ductal carcinoma": "Invasive Ductal"}
df["subtype"] = df["subtype"].replace(subtype_mapping)    

# er_status: ['Positive' 'ER+' 'Negative' 'ER positive' '+' 'ER-']
er_mapping = {"Positive": "ER+", "Negative": "ER-", "ER positive": "ER+", "+": "ER+"}
df["er_status"] = df["er_status"].replace(er_mapping)

# pr_status: ['Positive' 'PR+' 'PR positive' '+' 'Negative' 'PR-']
pr_mapping = {"Positive": "PR+", "PR positive": "PR+", "+": "PR+", "Negative": "PR-"}
df["pr_status"] = df["pr_status"].replace(pr_mapping)

# map her2_status to a standard format
her2_mapping = {"Negative": "HER2-", "HER2 negative": "HER2-", "-": "HER2-", "Positive": "HER2+"}
df["her2_status"] = df["her2_status"].replace(her2_mapping)
```

```python
# check values to confirm that they are consistent now
for column in df.columns:
    print(f"Unique values in {column}: {df[column].unique()}")
```

```python
# handle values in tumour_size
tumour_size_strip = df['tumour_size'].str.replace(' ', '')

# create tumour_size_unit column and set default to 'mm'
df['tumour_size_unit'] = 'mm'
df.loc[tumour_size_strip.str.endswith('cm'), 'tumour_size_unit'] = 'cm'  #if tumour_size ends with 'cm', set unit to 'cm'

# create tumour_size_number column by stripping 'cm' and 'mm' from tumour_size and converting to float
df['tumour_size_number'] = tumour_size_strip.str.rstrip('cm').str.rstrip('mm').astype(float)

# convert tumour_size_number to mm if tumour_size_unit is 'cm' in standardized_tumour_size_mm column
df['standardized_tumour_size_mm'] = df['standardized_tumour_size_mm'] = np.where(
    df['tumour_size_unit'] == 'cm', 
    df['tumour_size_number'] * 10, 
    df['tumour_size_number']
)

# print a subset of tumour size columns to check if the conversion worked
print(df[['tumour_size', 'tumour_size_number', 'tumour_size_unit', 'standardized_tumour_size_mm']].head())
```

```python
#filter dataset
filtered_df = df[df["her2_status"] == "HER2-"]

#check dataset
print(filtered_df.head())
```

```python
#export dataset for submission!
filtered_df.to_csv("challenge1_submission.csv", index=False)

print("Challenge 1 done!")
```
