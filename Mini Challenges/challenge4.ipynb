## Mini Challenge #4: Preprocessing

### Scenario

A surgical pathologist has collected patient reports from multiple hospitals. The dataset contains inconsistent terminology, spelling mistakes, missing values, duplicate records, and inconsistent measurements.

### Task

Preprocess the dataset so that it is clean and consistent. Your tasks are to:

1. Standardize variable names and terminology.
2. Correct inconsistent staging and grading formats.
3. Handle missing data.
4. Identify and remove duplicate records.
5. Standardize inconsistent measurements where needed.

### **note: you can subset the data for easier handling**

```python
# import libraries
import pandas as pd
```

```python
# check data for preprocessing steps that need to be applied
df = pd.read_csv("chellenge4_data.csv")  # replace only if the filename differs

print(df.head())
print(df.columns)
print(df.isnull().sum())
print(df.duplicated().sum())
```

```python
# handle missing values
df = df.dropna(subset=["Sample Id", "Sample Stage", "Sample Grade"]) #TODO: replace with specific columns

# handle duplicates by removing duplicate rows
df.drop_duplicates(inplace=True)
```

```python
# handle missing values
df = df.dropna(subset=["Sample Id", "Sample Stage", "Sample Grade"])

# standardize staging and grading terminology
df["Sample Stage"] = df["Sample Stage"].replace({
    "2": "II",
    "II": "II",
    "3": "III",
    "III": "III",
    "4": "IV",
    "IV": "IV"
})

df["Sample Grade"] = df["Sample Grade"].replace({
    "High": "High Grade",
    "high grade": "High Grade",
    "Low": "Low Grade",
    "low grade": "Low Grade"
})

# handle duplicates
df.drop_duplicates(inplace=True)
```

```python
# check cleaned dataset
print(df.head())
print(df.isnull().sum())
print(df.duplicated().sum())
```

```python
# export dataset for submission
df.to_csv("challenge4_submission.csv", index=False)
```
