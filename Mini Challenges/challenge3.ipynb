## Mini Challenge #3: Data Analysis & Visualization
### Scenario
The oncologist suspects that patients diagnosed at later stages tend to have higher recurrence rates. She also wants to investigate whether recurrence rates differ across age groups. She would like a figure suitable for inclusion in a presentation to other doctors. 
### Task
Given this cleaned dataset, your task is to construct a figure for this analysis. 

### **note: consider the most important variables in her research**

```python
# import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# assess data for relevant variables
df = pd.read_csv("chellenge3_data.csv") #TODO: replace with correct file
print(df.head())
print(df.columns)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6)) #TODO: edit sample stage column to be stage at diagnosis

ct_stage_rec = pd.crosstab(df["Stage At Diagnosis"], df["Recurrent Tumors"], normalize="index")
ct_stage_rec.plot(kind="bar", stacked=True, ax=axes[0], colormap="Blues")
axes[0].title("Recurrence by Stage at Diagnosis (normalized)")
axes[0].xlabel("Stage at Diagnosis")
axes[0].ylabel("Proportion")
axes[0].xticks(rotation=45)
axes[0].legend(title="Recurrent", bbox_to_anchor=(1.05, 1), loc="upper left")


bins = [0, 40, 50, 60, 70, 80, 100]
labels = ["<40", "40-49", "50-59", "60-69", "70-79", "80+"]
df["Age Group"] = pd.cut(df["Patient Age Diagnosis"], bins=bins, labels=labels, right=False)

ct_age_rec = pd.crosstab(df["Age Group"], df["Recurrent Tumors"], normalize="index")
ct_age_rec.plot(kind="bar", stacked=True, ax=axes[1], colormap="Blues")
axes[1].title("Recurrence by Age Group at Diagnosis (normalized)")
axes[1].xlabel("Age at Diagnosis")
axes[1].ylabel("Proportion")
axes[1].xticks(rotation=45)
axes[1].legend(title="Recurrent", bbox_to_anchor=(1.05, 1), loc="upper left")

plt.tight_layout()
plt.show()
```
