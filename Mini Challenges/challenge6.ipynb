# Mini Challenge #6: Survival Analysis

## Scenario

The pathologist wants to know whether different patient groups have different overall survival.

## Task

Using the cleaned patient dataset:

1. Create Kaplan–Meier survival curves grouped by stage.
2. Create Kaplan–Meier survival curves grouped by sex.
3. Add a short explanation of the results.

## Submission

Submit:

- One figure containing two Kaplan–Meier plots.
- A brief explanation of the observed survival patterns.

```python
# import libraries
import pandas as pd
import matplotlib.pyplot as plt
from lifelines import KaplanMeierFitter
```

```python
# load the cleaned dataset
df = pd.read_csv("challenge6.csv")  # replace if the filename differs

print(df.head())
print(df.columns)
```

```python
# Kaplan–Meier survival curves

time_column = "Overall Survival"
event_column = "Death"
stage_column = "Stage At Diagnosis"
sex_column = "Sex"

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Survival by stage
for group in df[stage_column].dropna().unique():
    group_data = df[df[stage_column] == group]

    kmf = KaplanMeierFitter()
    kmf.fit(
        durations=group_data[time_column],
        event_observed=group_data[event_column],
        label=str(group)
    )
    kmf.plot_survival_function(ax=axes[0])

axes[0].set_title("Overall Survival by Stage")
axes[0].set_xlabel("Time")
axes[0].set_ylabel("Survival Probability")
axes[0].legend(title="Stage")

# Survival by sex
for group in df[sex_column].dropna().unique():
    group_data = df[df[sex_column] == group]

    kmf = KaplanMeierFitter()
    kmf.fit(
        durations=group_data[time_column],
        event_observed=group_data[event_column],
        label=str(group)
    )
    kmf.plot_survival_function(ax=axes[1])

axes[1].set_title("Overall Survival by Sex")
axes[1].set_xlabel("Time")
axes[1].set_ylabel("Survival Probability")
axes[1].legend(title="Sex")

plt.tight_layout()
plt.show()
```

Explanation:

The Kaplan–Meier curves show the estimated probability of surviving over time. Groups with curves that remain higher have better observed overall survival. Patients whose curves fall more quickly have poorer survival. Downward steps represent observed deaths, while flat sections indicate periods where no deaths were observed.
